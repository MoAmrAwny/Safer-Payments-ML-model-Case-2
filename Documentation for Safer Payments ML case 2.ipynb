## Preprocessing and EDA

### Exploring and understading the data 

* First, we needed to know the data type of our features and how many null values are there, then we found that there isn't many null values so we dropped all the null values.



* After carefully studying our Dataset we found that we have 2 columns that contain dates, the first is DateOfBirth and the second is TransactionDate. We agreed that we don't need the whole date for these 2 columns, So acccordingly we agreed that we only needed the Year for the DateOfBirth and the Month for the TransactionDate.

    **So** we created 2 columns and named them Year and TransactionMonth and dropped DateOfBirth and TransactionDate as they are not relevant anymore.


* We noticed that there are 5 accounts for each client with 5 different transactions and 5 different account balances, 
    
    **So** we decided to merge the 5 account balances into only one account and drop the other 4, and this will be our Data that we will work with.

### Encoding

## Feature Selection 

* firstly we needed to Standardize the features, so that they have a mean of 0 and a standard deviation of 1. This ensures that each feature contributes equally to the analysis. So we used StandardScaler

* Then we used the Variance Threshold method to select our most important features.

* Compute correlation matrix and created heatmap and removed highly correlated features(above 0.7).

* Then we created a new dataset with only the important features and filtered2_df

## Training the Model

* First we used the Elbow Method to determine the optimum number of Clusters to use (as Kmeans and Guassian Mixture both require Predefined number of Clusters).

* Then we used Kmeans algorithm to perform clustering, because it is computationally efficient and provides straightforward, interpretable results. For clustering clients into predefined tiers, it also helps in assigning each client to one of the distinct tiers based on their principal components.
 
  **Silhoutte Score** was 0.1474557901099864


* Then we used Guassian Mixture algorithm to perform clustering, can handle clusters of varying shapes and sizes by assuming that the data is generated from a mixture of several Gaussian distributions. This flexibility can be beneficial if the clusters have overlapping boundaries or varying densities, which might be the case with bank clients who exhibit diverse account behaviors.
 
  **Silhoutte Score** was 0.09542910950783756
  

* Then we used PCA algorithm to perform clustering, helps in visualizing high-dimensional data and reduces noise by focusing on principal components that capture the most significant variance.

  **Silhoutte Score** was 0.3632358003849705
